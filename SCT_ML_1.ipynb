## TASK 01
Implement a linear regression model to predict the prices of houses based on their square footage and the number of bedrooms and bathrooms.

```python
## Reading the data using pandas
import numpy as np
import pandas as pd
df = pd.read_csv('train.csv')
```

```python
df.head()
```

```python
df.columns.tolist()
```

```python
df_model = df[['BedroomAbvGr','FullBath','GrLivArea','SalePrice']].dropna().copy()
## converting into numpy arrays
X_raw = df_model[['BedroomAbvGr','FullBath','GrLivArea']].values
y = df_model['SalePrice'].values
X_raw.shape
y.shape
```

```python
ones = np.ones((X_raw.shape[0],1))
X = np.hstack([ones,X_raw])
print("X with bias shape:",X.shape)
```

```python
np.random.seed(42)
indices = np.random.permutation(len(X))
split = int(0.8*len(X))
train_idx = indices[:split]
test_idx = indices[split:]
X_train = X[train_idx]
X_test = X[test_idx]
Y_train = y[train_idx]
Y_test = y[test_idx]
```

```python
XT = X_train.T
XTX = np.dot(XT,X_train)
XTX_inv = np.linalg.inv(XTX)
XTy = np.dot(XT,Y_train)
beta = np.dot(XTX_inv,XTy)
beta
```

```python
y_pred = np.dot(X_test,beta)
```

```python
for i in range(5):
    print(f"Actual: {Y_test[i]:.0f}  |  Predicted: {y_pred[i]:.0f}")
```

```python
errors = Y_test - y_pred
rmse = np.sqrt(np.mean(errors**2))
print("RMSE:", round(rmse,2))
```

```python
ss_res = np.sum((Y_test - y_pred) ** 2)        # residual sum of squares
ss_tot = np.sum((Y_test - np.mean(Y_test))**2)  # total sum of squares
r2 = 1 - (ss_res / ss_tot)
print("R² Score:", round(r2, 4))
```
